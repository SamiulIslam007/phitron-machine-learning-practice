```python
def make_prediction(X, y, W, b):
    m = X.shape[0] # Number of training examples
    pred_list = np.zeros((m,)) #Initialize an array to store results


    for i in range(m):
        #X[i] = feature vector , W = weights_vector
        pred_list[i] = np.dot(W, X[i]) + b



    return pred_list
```

```python
m = X.shape[0]
n = X.shape[1]

W_init = np.ones((n,))
b_init = 1.0

make_prediction(X, y , W_init, b_init)
```

```python
def compute_cost(X, y, W,b):
    m = X.shape[0] # numbers of data points

    cost = 0.0 

    pred_list = make_prediction(X, y, W, b)
    # print(pred_list)

    error = pred_list - y
    # print(error)

    error_squared = error ** 2
    # print(error_squared)

    cost = np.sum(error_squared)

    #  We divide by 2m for mathematical convenience in gradient calculations
    cost = cost / (2 * m)
    return cost
```

```python
W_init = np.ones((n,))
b_init = 1.0 
compute_cost(X, y, W_init, b_init) 
```

```python
def calculate_gradient(X, y, W, b):
    """Computes the gradient (direction to move) for w and b"""
    m = X.shape[0]
    n = X.shape[1]
    
    dj_dw = np.zeros((n,)) # joto ta feature ache
    dj_db = 0.0 # Change needed for bias
    
    for i in range(m):
        prediction = np.dot(W, X[i]) + b
        error = prediction - y[i]

        dj_db = dj_db + error
        
        for j in range(n):
            dj_dw[j] = dj_dw[j] + (error * X[i,j])


    return dj_dw / m , dj_db / m          
```

```python
W_init = np.ones((n,))
b_init = 1.0 
calculate_gradient(X, y, W_init, b_init) 
```
