```python
x = 12
x = x/100
min_range = df['maths_marks'].quantile(x)
max_range = df['maths_marks'].quantile(1-x)

print(f'min range : {min_range} and max range:{max_range}')
```

```python
df['maths_marks'] = df['maths_marks'].clip(min_range,max_range)

```
